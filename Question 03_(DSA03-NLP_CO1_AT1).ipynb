```python
import re


def validate_register_number(register_number):
    pattern = r"^\d{4}[A-Z]{2,4}\d{3}$"
    return bool(re.fullmatch(pattern, register_number))


def validate_email(email):
    pattern = r"^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.edu$"
    return bool(re.fullmatch(pattern, email))


def validate_course_code(course_code):
    pattern = r"^[A-Z]{3}\d{2}$"
    return bool(re.fullmatch(pattern, course_code))


def validate_semester(semester):
    pattern = r"^Semester [1-8]$"
    return bool(re.fullmatch(pattern, semester))


def validate_mobile(mobile):
    pattern = r"^[6-9]\d{9}$"
    return bool(re.fullmatch(pattern, mobile))


print("=" * 60)
print("UNIVERSITY REGISTRATION VALIDATION SYSTEM")
print("=" * 60)

register_number = input("Enter Register Number: ").strip()
email = input("Enter Institutional Email: ").strip()
course_code = input("Enter Course Code: ").strip()
semester = input("Enter Semester: ").strip()
mobile = input("Enter Mobile Number: ").strip()

print("\n" + "=" * 60)
print("VALIDATION RESULTS")
print("=" * 60)

all_valid = True

if validate_register_number(register_number):
    print("Register Number       : VALID")
else:
    print("Register Number       : INVALID")
    all_valid = False


if validate_email(email):
    print("Institutional Email   : VALID")
else:
    print("Institutional Email   : INVALID")
    all_valid = False


if validate_course_code(course_code):
    print("Course Code           : VALID")
else:
    print("Course Code           : INVALID")
    all_valid = False


if validate_semester(semester):
    print("Semester              : VALID")
else:
    print("Semester              : INVALID")
    all_valid = False


if validate_mobile(mobile):
    print("Mobile Number         : VALID")
else:
    print("Mobile Number         : INVALID")
    all_valid = False


print("\n" + "=" * 60)

if all_valid:
    print("FINAL REGISTRATION STATUS : SUCCESSFUL")
    print("All student details are valid.")
else:
    print("FINAL REGISTRATION STATUS : UNSUCCESSFUL")
    print("Please correct the invalid details.")

print("=" * 60)
```
