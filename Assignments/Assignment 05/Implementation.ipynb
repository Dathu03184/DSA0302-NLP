TASK 1 – Discourse and the Unit of Analysis (3 Marks)

```python
import re

report = """
The old suspension bridge near Periyar river was closed at 6 AM.
A state transport bus carrying twenty-five passengers attempted to cross the northern causeway.
It collapsed into the raging waters five minutes ago.
"""

def segment_discourse(text):
    sentences = re.split(r'(?<=[.!?])\s+', text.strip())

    for i, sentence in enumerate(sentences, 1):
        if sentence:
            print("EDU", i, ":", sentence)

print("DISCOURSE SEGMENTATION")
segment_discourse(report)
```

Task 2 – Reference Resolution (5 Marks)

```python
import re

text = [
    "The bridge was damaged by heavy rain.",
    "The rescue team reached the bridge.",
    "It was closed for public safety."
]

def reference_resolution(sentences):
    nouns = []

    for sentence in sentences:
        words = re.findall(r'\b[A-Z][a-z]+\b|\bbridge\b|\bteam\b', sentence)
        pronouns = re.findall(r'\b(It|They|He|She|This|That)\b', sentence)

        for pronoun in pronouns:
            if nouns:
                print(pronoun, "may refer to:", nouns[-1])

        nouns.extend(words)

reference_resolution(text)

def centering_theory():
    sentences = [
        ["bridge", "rain"],
        ["team", "bridge"],
        ["bridge", "safety"]
    ]

    previous = None

    for sentence in sentences:
        current = sentence[0]

        if previous is None:
            transition = "Start"
        elif current == previous:
            transition = "Continue"
        elif previous in sentence:
            transition = "Retain"
        else:
            transition = "Shift"

        print("Center:", current, "Transition:", transition)
        previous = current

centering_theory()
```

Task 3 – Text Coherence (4 Marks)

```python
sentences = [
    ["bridge", "rain"],
    ["bridge", "rescue"],
    ["rescue", "people"],
    ["people", "shelter"]
]

def entity_grid(data):
    for i, sentence in enumerate(data, 1):
        print("Sentence", i, ":", sentence)

def coherence_score(data):
    score = 0

    for i in range(len(data) - 1):
        common = set(data[i]) & set(data[i + 1])
        score += len(common)

    return score

entity_grid(sentences)

print("Coherence Score:", coherence_score(sentences))
```

Task 5 – Dialog and Conversational Agents (4 Marks)

```python
class ReliefCampAgent:

    def __init__(self):
        self.state = "START"
        self.location = ""

    def respond(self, message):
        message = message.lower()

        if message == "cancel":
            self.state = "END"
            return "Request cancelled."

        if self.state == "START":
            self.state = "LOCATION"
            return "Please enter your location."

        elif self.state == "LOCATION":
            self.location = message
            self.state = "CONFIRM"
            return "Your location is " + self.location + ". Is this correct?"

        elif self.state == "CONFIRM":
            if message == "yes":
                self.state = "END"
                return "Finding the nearest relief camp."
            else:
                self.state = "LOCATION"
                return "Please enter your location again."

agent = ReliefCampAgent()

print(agent.respond("hello"))
print(agent.respond("Aluva"))
print(agent.respond("yes"))
```

Task 6 – Dialog Acts and Interpretation (4 Marks)

```python
def dialogue_act(message):
    message = message.lower()

    if any(word in message for word in ["where", "which", "when"]):
        return "Question"

    elif any(word in message for word in ["help", "need", "require"]):
        return "Request"

    elif any(word in message for word in ["yes", "correct", "okay"]):
        return "Confirmation"

    elif any(word in message for word in ["no", "wrong"]):
        return "Correction"

    elif any(word in message for word in ["thank"]):
        return "Thanking"

    else:
        return "Statement"

messages = [
    "Where is the nearest shelter?",
    "I need help.",
    "Yes, that is correct.",
    "No, that is wrong.",
    "Thank you."
]

for message in messages:
    print(message, ":", dialogue_act(message))
```

Task 9 – Machine Translation of the Bulletins (5 Marks)

```python
import math

source = ["the road is closed", "the bridge is damaged"]
target = ["sadak band hai", "pul kshatigrast hai"]

dictionary = {
    "the": "the",
    "road": "sadak",
    "is": "hai",
    "closed": "band",
    "bridge": "pul",
    "damaged": "kshatigrast"
}

def direct_translation(sentence):
    words = sentence.lower().split()
    return " ".join(dictionary.get(word, word) for word in words)

print("DIRECT TRANSLATION")

for sentence in source:
    print(direct_translation(sentence))

def transfer_translation(sentence):
    translated = direct_translation(sentence)
    return translated

print("\nTRANSFER TRANSLATION")
print(transfer_translation("the road is closed"))

interlingua = {
    "event": "ROAD_CLOSURE",
    "object": "ROAD",
    "status": "CLOSED"
}

print("\nINTERLINGUA REPRESENTATION")
print(interlingua)

def ibm_model1(source_sentences, target_sentences, iterations=5):
    pairs = []

    for s, t in zip(source_sentences, target_sentences):
        pairs.append((s.split(), t.split()))

    vocabulary = set()

    for s, t in pairs:
        for word in s:
            for target_word in t:
                vocabulary.add((word, target_word))

    probability = {pair: 1 / 10 for pair in vocabulary}

    for iteration in range(iterations):
        count = {}
        total = {}

        for source_words, target_words in pairs:
            for target_word in target_words:
                denominator = sum(
                    probability.get((word, target_word), 0.0001)
                    for word in source_words
                )

                for word in source_words:
                    key = (word, target_word)
                    value = probability.get(key, 0.0001) / denominator
                    count[key] = count.get(key, 0) + value
                    total[word] = total.get(word, 0) + value

        for key in count:
            probability[key] = count[key] / total[key[0]]

        print("Iteration:", iteration + 1, "Completed")

    return probability

model = ibm_model1(source, target)

print("\nTRANSLATION PROBABILITIES")

for key, value in model.items():
    print(key, ":", round(value, 3))
```

Task 10 – Implement and Test the System (3 Marks)

```python
import time

def test_segmenter():
    text = "The road is closed. People need help."
    result = text.split(".")
    return len([x for x in result if x.strip()]) == 2

def test_reference_resolution():
    sentences = [
        "The bridge is damaged.",
        "It is closed."
    ]
    return "It" in sentences[1]

def test_coherence():
    data = [
        ["bridge", "rain"],
        ["bridge", "rescue"]
    ]
    return len(set(data[0]) & set(data[1])) == 1

def test_dialogue():
    return dialogue_act("Where is the shelter?") == "Question"

def test_generation():
    result = surface_realisation(
        "Aluva", "Flood", "Periyar Road", "Government School"
    )
    return len(result) == 3

tests = [
    ("Discourse Segmentation", test_segmenter),
    ("Reference Resolution", test_reference_resolution),
    ("Text Coherence", test_coherence),
    ("Dialogue Act", test_dialogue),
    ("Language Generation", test_generation)
]

print("SYSTEM TESTING RESULTS")

start = time.time()

for name, test in tests:
    result = test()
    print(name, ":", "PASS" if result else "FAIL")

end = time.time()

print("Execution Time:", round(end - start, 5), "seconds")
```

Task 12 – Re-evaluate and Adapt the Design (8 Marks)

```python
import re

messages = [
    "Person A: The bridge is damaged.",
    "Person B: Road block hai near the bridge.",
    "Person C: It is not safe to travel.",
    "Person D: Please send help."
]

def clean_message(message):
    message = message.lower()
    message = re.sub(r'[^a-z0-9\s:]', '', message)
    return message.strip()

def detect_language_type(message):
    if any(word in message for word in ["hai", "nahi", "kya"]):
        return "Code-Mixed"
    return "English"

def process_group_messages(messages):
    for message in messages:
        cleaned = clean_message(message)
        language = detect_language_type(cleaned)

        print("\nOriginal:", message)
        print("Cleaned:", cleaned)
        print("Language:", language)
        print("Dialogue Act:", dialogue_act(cleaned))

def speech_correction(text):
    corrections = {
        "brige": "bridge",
        "floodin": "flooding",
        "sheltr": "shelter",
        "rood": "road"
    }

    words = text.lower().split()
    corrected = [corrections.get(word, word) for word in words]

    return " ".join(corrected)

print("GROUP MESSAGE PROCESSING")
process_group_messages(messages)

print("\nSPEECH RECOGNITION CORRECTION")
noisy_text = "The brige is damaged near the rood"
print("Original:", noisy_text)
print("Corrected:", speech_correction(noisy_text))

print("\nADAPTED SYSTEM")
print("1. Use speaker identification.")
print("2. Separate reports by incident.")
print("3. Apply code-mixed language processing.")
print("4. Correct speech recognition errors.")
print("5. Use multilingual bulletin generation.")
print("6. Verify emergency information before publication.")
```
