Task 01 – Meaning Representation

```python
def canonicalise(text):
    text = text.lower()

    synonyms = {
        "retiree": "pensioner",
        "lump sum": "commutation",
        "withdraw upfront": "commutation",
        "years worked": "qualifying service"
    }

    for word, meaning in synonyms.items():
        text = text.replace(word, meaning)

    return text


q1 = "What is the lump sum for a retiree?"
q2 = "What is the commutation for a pensioner?"

print(canonicalise(q1))
print(canonicalise(q2))
```

Task 02 – Meaning Structure of Language

```python
def meaning_check(statement):
    knowledge_base = {
        "minimum service is 10 years": True,
        "pension is available after 10 years": True,
        "all employees receive pension": None
    }

    return knowledge_base.get(statement, "UNKNOWN")


print(meaning_check("minimum service is 10 years"))
print(meaning_check("pension is available after 10 years"))
print(meaning_check("all employees receive pension"))
```

Task 08 – Information Retrieval using TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Qualifying service and minimum ten years pension",
    "Family pension for widowed daughters",
    "Revision of pension commutation factor",
    "Military service towards civil pension",
    "Restoration of commuted pension"
]

query = ["qualifying service for pension"]

vectorizer = TfidfVectorizer()

doc_vectors = vectorizer.fit_transform(documents)
query_vector = vectorizer.transform(query)

similarity = cosine_similarity(query_vector, doc_vectors)

for i, score in enumerate(similarity[0]):
    print("Document", i + 1, "Similarity:", round(score, 4))

ranking = similarity[0].argsort()[::-1]

print("\nRanked Documents:")

for index in ranking:
    print(documents[index])
```

Task 09 – Cohen's Kappa

```python
from sklearn.metrics import cohen_kappa_score

annotator1 = [
    "employment",
    "military",
    "pension",
    "family",
    "benefit"
]

annotator2 = [
    "employment",
    "military",
    "pension",
    "family",
    "assertion"
]

agreement = sum(
    a == b for a, b in zip(annotator1, annotator2)
) / len(annotator1)

kappa = cohen_kappa_score(annotator1, annotator2)

print("Agreement:", agreement)
print("Cohen's Kappa:", kappa)
```

Task 10 – Implement and Test the Complete Pipeline

```python
import time
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity

documents = [
    "Qualifying service minimum ten years pension",
    "Family pension for dependent daughters",
    "Revision of commutation factor",
    "Military service towards civil pension",
    "Restoration of commuted pension"
]

queries = [
    "qualifying service pension",
    "family pension daughter",
    "commutation factor revision"
]

start = time.time()

vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(documents)

for query in queries:
    query_vector = vectorizer.transform([query])
    scores = cosine_similarity(query_vector, matrix)[0]

    print("\nQuery:", query)

    for index in scores.argsort()[::-1]:
        print(documents[index], round(scores[index], 3))

end = time.time()

print("\nTotal Runtime:", round(end - start, 4), "seconds")
```

Task 12 – Re-evaluate and Adapt the Design

```python
def calculate_service(start_year, end_year, break_years):
    total_years = end_year - start_year + 1
    qualifying_years = total_years - break_years

    return qualifying_years


def pension_eligibility(retirement_year, service_years):
    if retirement_year < 2024:
        minimum_years = 10
    else:
        minimum_years = 25

    if service_years >= minimum_years:
        return "Eligible for pension"
    else:
        return "Not eligible for pension"


start_year = 1995
end_year = 2019
break_years = 2

service = calculate_service(
    start_year,
    end_year,
    break_years
)

print("Qualifying Service:", service, "years")

result = pension_eligibility(2019, service)

print("Result:", result)
```
