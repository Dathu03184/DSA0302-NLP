```python
import re

products = [
    "Gaming Laptop",
    "Laptop Stand",
    "Smart Phone",
    "Smart Watch",
    "Smart TV",
    "Wireless Mouse",
    "Wireless Keyboard",
    "Bluetooth Speaker",
    "Notebook",
    "Book Stand",
    "Phone Case",
    "Laptop Bag"
]


def exact_search(keyword):
    pattern = r"\b" + re.escape(keyword) + r"\b"

    return [
        product for product in products
        if re.search(pattern, product, re.IGNORECASE)
    ]


def prefix_search(keyword):
    pattern = r"^" + re.escape(keyword)

    return [
        product for product in products
        if re.search(pattern, product, re.IGNORECASE)
    ]


def suffix_search(keyword):
    pattern = re.escape(keyword) + r"$"

    return [
        product for product in products
        if re.search(pattern, product, re.IGNORECASE)
    ]


def partial_search(keyword):
    pattern = re.escape(keyword)

    return [
        product for product in products
        if re.search(pattern, product, re.IGNORECASE)
    ]


def display_results(search_type, results):

    print("\n" + search_type)
    print("-" * 40)

    if results:
        for product in results:
            print(product)

        print("Total matching products:", len(results))
    else:
        print("No products found.")
        print("Total matching products: 0")


print("=" * 50)
print("PRODUCT SEARCH SYSTEM")
print("=" * 50)

keyword = input("Enter search keyword: ").strip()

if keyword == "":
    print("Search keyword cannot be empty.")

else:

    exact_results = exact_search(keyword)
    display_results("EXACT SEARCH", exact_results)

    prefix_results = prefix_search(keyword)
    display_results("PREFIX SEARCH", prefix_results)

    suffix_results = suffix_search(keyword)
    display_results("SUFFIX SEARCH", suffix_results)

    partial_results = partial_search(keyword)
    display_results("PARTIAL SEARCH", partial_results)
```
