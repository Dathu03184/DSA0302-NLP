```python
from collections import Counter
corpus = ["i like tea", "i like coffee", "i drink tea"]
bigrams = Counter()
words = Counter()
vocab = set()
for sentence in corpus:
    w = sentence.split()
    vocab.update(w)
    for i in range(len(w) - 1):
        bigrams[(w[i], w[i+1])] += 1
        words[w[i]] += 1
v = len(vocab)

pairs = [("i", "like"), ("like", "tea"),
         ("like", "juice"), ("i", "drink")]
for w1, w2 in pairs:
    count = bigrams[(w1, w2)]
    mle = count / words[w1] if words[w1] else 0
    smooth = (count + 1) / (words[w1] + v)
    print(w1, w2)
    print("MLE:", round(mle, 3))
    print("Add-one:", round(smooth, 3))
```
