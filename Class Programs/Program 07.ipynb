```python
suffixes = {
    "ization": "Noun",
    "ation": "Noun",
    "ness": "Noun",
    "ment": "Noun",
    "tion": "Noun",
    "er": "Noun",
    "ly": "Adverb",
    "ize": "Verb",
    "ise": "Verb",
    "al": "Adjective",
    "ful": "Adjective",
    "less": "Adjective",
    "able": "Adjective"
}

word = input("Enter a word: ").lower()

found = False

for suffix, word_class in suffixes.items():
    if word.endswith(suffix) and len(word) > len(suffix):
        base = word[:-len(suffix)]

        print("\nWord:", word)
        print("Base word:", base)
        print("Derivational suffix:", suffix)
        print("New word class:", word_class)
        print("Meaning/word class changes after adding the suffix.")

        found = True
        break

if not found:
    print("\nNo derivational suffix found.")
```
