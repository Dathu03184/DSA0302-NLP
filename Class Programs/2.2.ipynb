```python
import random
from collections import Counter
corpus = ["i like tea", "i like coffee", "you like tea"]
bigrams = Counter()
words_count = Counter()
model = {}
for s in corpus:
    words = ["<s>"] + s.split() + ["</s>"]
    for i in range(len(words)-1):
        bigrams[(words[i], words[i+1])] += 1
        words_count[words[i]] += 1
        model.setdefault(words[i], []).append(words[i+1])
print("Bigram Probabilities:")
for (w1, w2), count in bigrams.items():
    print(w1, w2, "=", round(count / words_count[w1], 2))
s = ["<s>"] + input("Enter sentence: ").split() + ["</s>"]
p = 1
for i in range(len(s)-1):
    p *= bigrams[(s[i], s[i+1])] / words_count[s[i]] if words_count[s[i]] else 0
print("Sentence Probability:", round(p, 4))
print("Generated Sentences:")
for _ in range(3):
    w = "<s>"
    result = []
    while w != "</s>":
        w = random.choice(model[w])
        if w != "</s>":
            result.append(w)
    print(" ".join(result))
```
