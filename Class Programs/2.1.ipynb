```python

from collections import Counter

# Part 1: Print unigram, bigram and trigram
sentence = input("Enter a sentence: ").lower()
words = sentence.split()

unigrams = words
bigrams = list(zip(words, words[1:]))
trigrams = list(zip(words, words[1:], words[2:]))

print("Unigrams:", unigrams)
print("Bigrams:", bigrams)
print("Trigrams:", trigrams)

# Part 2: Predict the next word
corpus = "i like tea i like coffee i like tea"

corpus_words = corpus.split()
bigram_counts = Counter(
    zip(corpus_words, corpus_words[1:])
)

word = input("Enter a word to predict next word: ").lower()

candidates = {
    next_word: count
    for (current_word, next_word), count in bigram_counts.items()
    if current_word == word
}

if candidates:
    prediction = max(candidates, key=candidates.get)
    print("Next word:", prediction)
else:
    print("Word not found in corpus")
```
