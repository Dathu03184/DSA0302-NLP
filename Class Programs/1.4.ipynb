```python
import re

patterns = {
    "?": r"colou?r",
    "*": r"ab*c",
    "+": r"ab+c",
    "[]": r"[aeiou]",
    "^": r"^hello",
    "$": r"world$"
}

test_strings = [
    "color",
    "colour",
    "ac",
    "abc",
    "abbc",
    "apple",
    "sky",
    "hello there",
    "world",
    "my world"
]

for operator, pattern in patterns.items():
    print("\nTesting operator:", operator)
    print("Pattern:", pattern)

    for s in test_strings:
        if re.search(pattern, s):
            print("String:", s, "-> YES")
        else:
            print("String:", s, "-> NO")
```
