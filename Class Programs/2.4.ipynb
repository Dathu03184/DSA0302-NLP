```python
from collections import Counter

corpus = ["i like tea", "i like coffee", "you like tea"]

uni = Counter()
bi = Counter()
tri = Counter()

for s in corpus:
    w = s.split()
    uni.update(w)
    bi.update(zip(w, w[1:]))
    tri.update(zip(w, w[1:], w[2:]))

total = sum(uni.values())

a, b, c = input("Enter 3 words: ").lower().split()

if tri[(a, b, c)] > 0:
    p = tri[(a, b, c)] / bi[(a, b)]
    print("Trigram probability:", p)
    print("Level: Trigram")
elif bi[(b, c)] > 0:
    p = (bi[(b, c)] / uni[b]) * 0.4
    print("Bigram backoff probability:", p)
    print("Level: Bigram")
elif uni[c] > 0:
    p = (uni[c] / total) * 0.4 * 0.4
    print("Unigram backoff probability:", p)
    print("Level: Unigram")
else:
    print("Sequence not found")
```
