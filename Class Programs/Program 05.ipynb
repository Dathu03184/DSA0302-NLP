```python
def check_string(s):
    state = 'q0'
    path = ['q0']

    for ch in s:
        if state == 'q0':
            state = 'q1' if ch == 'a' else 'q0'

        elif state == 'q1':
            state = 'q1' if ch == 'a' else 'q2'

        elif state == 'q2':
            state = 'q1' if ch == 'a' else 'q0'

        path.append(state)

    if state == 'q2':
        print(f"String: {s} -> Path: {' -> '.join(path)} -> Accepted")
    else:
        print(f"String: {s} -> Path: {' -> '.join(path)} -> Rejected")


# Test strings
test_strings = ["ab", "aab", "b", "aba", "abb", "abab", "ba"]

for s in test_strings:
    check_string(s)
```
