```python
def dialogue_act(text):
    text = text.lower().strip()

    if any(word in text for word in ["hello", "hi", "hey"]):
        return "Greeting"
    elif text.endswith("?"):
        return "Question"
    elif text.startswith("please"):
        return "Request"
    else:
        return "Statement"

text = input("Enter dialogue: ")
print("Dialogue act:", dialogue_act(text))
```
