```python
import nltk
from nltk.wsd import lesk

nltk.download("wordnet")
nltk.download("omw-1.4")

sentence = "He deposited money in the bank".split()

meaning = lesk(sentence, "bank")

if meaning:
    print("Word: bank")
    print("Meaning:", meaning.definition())
else:
    print("Meaning not found")
```
