```python
import nltk
from nltk.wsd import lesk

nltk.download('wordnet')

sentence = "He deposited money in the bank"
words = sentence.split()

result = lesk(words, 'bank')

if result:
    print("Word:", "bank")
    print("Meaning:", result.definition())
else:
    print("Meaning not found")
```
