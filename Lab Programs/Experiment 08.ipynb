```python
import random

tag_model = {
    "the": {"DT": 0.9, "NN": 0.1},
    "boy": {"NN": 0.9, "VB": 0.1},
    "plays": {"VB": 0.8, "NNS": 0.2},
    "cricket": {"NN": 1.0}
}

sentence = "the boy plays cricket".split()

for word in sentence:
    if word in tag_model:
        tags = list(tag_model[word].keys())
        probs = list(tag_model[word].values())
        tag = random.choices(tags, weights=probs)[0]
    else:
        tag = "NN"
    print(word, "->", tag)
```
