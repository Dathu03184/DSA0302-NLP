```python
import re

text = """Call me on 12/09/2026 or 25-10-2026.
My mobile numbers are 9876543210 and 9123456789.
Email me at example@gmail.com or student@gmail.com.
HELLO WORLD PYTHON NLP
Call me for more details."""

date_pattern = r'\b\d{1,2}[-/]\d{1,2}[-/]\d{4}\b'
first_date = re.search(date_pattern, text)

print("1. First date:")
print(first_date.group() if first_date else "No date found")

mobile_pattern = r'\b[6-9]\d{9}\b'
mobile_numbers = re.findall(mobile_pattern, text)

print("\n2. All mobile numbers:")
print(mobile_numbers)

email_pattern = r'\b[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}\b'
emails = re.findall(email_pattern, text)

print("\n3. All email IDs:")
print(emails)

capital_pattern = r'\b[A-Z]+\b'
capital_words = re.findall(capital_pattern, text)

print("\n4. All capital words:")
print(capital_words)

starts_with_call = re.match(r'^Call', text)

print("\n5. Does the text start with 'Call'?")
print("Yes" if starts_with_call else "No")

new_text = re.sub(
    r'(\d{1,2})[-/](\d{1,2})[-/](\d{4})',
    r'\3-\2-\1',
    text
)

print("\n6. Text after changing date format:")
print(new_text)

hidden_text = re.sub(
    r'\b[6-9]\d{9}\b',
    'XXXXXXXXXX',
    text
)

print("\n7. Text after hiding phone numbers:")
print(hidden_text)
```
