```python
import nltk
from nltk import PCFG

grammar = PCFG.fromstring("""
S -> NP VP [1.0]
NP -> 'John' [0.5] | 'Mary' [0.5]
VP -> 'runs' [0.7] | 'sleeps' [0.3]
""")

parser = nltk.ViterbiParser(grammar)

for tree in parser.parse("John runs".split()):
    print(tree)
    print("Probability:", tree.prob())
```
