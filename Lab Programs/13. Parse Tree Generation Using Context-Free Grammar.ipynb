```python
import nltk
from nltk import CFG

grammar = CFG.fromstring("""
S -> NP VP
NP -> 'the' 'cat'
VP -> 'sleeps'
""")

parser = nltk.ChartParser(grammar)

for tree in parser.parse("the cat sleeps".split()):
    print(tree)
    tree.pretty_print()
```
