```python
from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS

text = input("Enter a paragraph: ")
sentences = [s.strip() for s in text.split(".") if s.strip()]

def get_words(sentence):
    return {
        word.lower().strip(" ,!?")
        for word in sentence.split()
        if word.lower().strip(" ,!?") not in ENGLISH_STOP_WORDS
    }

scores = []

for i in range(len(sentences) - 1):
    first = get_words(sentences[i])
    second = get_words(sentences[i + 1])

    union = first | second
    overlap = first & second

    score = len(overlap) / len(union) if union else 0
    scores.append(score)
    print("Sentence pair", i + 1, ":", round(score, 2))

if scores:
    print("Average coherence score:", round(sum(scores) / len(scores), 2))
else:
    print("Enter at least two sentences.")
```
