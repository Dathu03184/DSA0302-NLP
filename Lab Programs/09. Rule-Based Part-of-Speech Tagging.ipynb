```python
import nltk
from nltk import RegexpTagger

patterns = [
    (r".*ing$", "VBG"),
    (r".*ed$", "VBD"),
    (r".*ly$", "RB"),
    (r".*s$", "NNS"),
    (r".*", "NN")
]

tagger = RegexpTagger(patterns)

words = "playing walked quickly cats boy".split()

print(tagger.tag(words))
```
