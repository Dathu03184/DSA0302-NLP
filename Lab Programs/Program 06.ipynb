```python
irregular_nouns = {
    "man": "men",
    "woman": "women",
    "child": "children",
    "foot": "feet",
    "tooth": "teeth",
    "mouse": "mice"
}

irregular_verbs = {
    "go": "went",
    "eat": "ate",
    "come": "came",
    "write": "wrote",
    "see": "saw",
    "run": "ran"
}

irregular_adjectives = {
    "good": ("better", "best"),
    "bad": ("worse", "worst"),
    "far": ("farther", "farthest")
}

noun = input("Enter a noun: ").lower()
verb = input("Enter a verb: ").lower()
adjective = input("Enter an adjective: ").lower()

# Plural noun
if noun in irregular_nouns:
    plural = irregular_nouns[noun]
elif noun.endswith(("s", "x", "ch", "sh")):
    plural = noun + "es"
elif noun.endswith("y") and noun[-2] not in "aeiou":
    plural = noun[:-1] + "ies"
else:
    plural = noun + "s"

# Past tense
if verb in irregular_verbs:
    past = irregular_verbs[verb]
elif verb.endswith("e"):
    past = verb + "d"
elif verb.endswith("y") and verb[-2] not in "aeiou":
    past = verb[:-1] + "ied"
else:
    past = verb + "ed"

# -ing form
if verb.endswith("e") and not verb.endswith("ee"):
    ing = verb[:-1] + "ing"
else:
    ing = verb + "ing"

# Comparative and superlative
if adjective in irregular_adjectives:
    comparative, superlative = irregular_adjectives[adjective]
elif len(adjective) <= 5:
    comparative = adjective + "er"
    superlative = adjective + "est"
else:
    comparative = "more " + adjective
    superlative = "most " + adjective

print("\nPlural:", plural)
print("Past tense:", past)
print("Ing form:", ing)
print("Comparative:", comparative)
print("Superlative:", superlative)
```
