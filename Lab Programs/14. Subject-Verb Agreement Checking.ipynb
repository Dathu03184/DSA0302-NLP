```python
import nltk
from nltk import CFG

grammar = CFG.fromstring("""
S -> SG_NP SG_VP
S -> PL_NP PL_VP
SG_NP -> 'he' | 'she'
PL_NP -> 'they' | 'we'
SG_VP -> 'runs' | 'sleeps'
PL_VP -> 'run' | 'sleep'
""")

parser = nltk.ChartParser(grammar)

sentence = input("Enter sentence: ").lower().split()

if list(parser.parse(sentence)):
    print("Correct agreement")
else:
    print("Incorrect agreement")
```
