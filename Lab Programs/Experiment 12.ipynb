```python
import nltk
from nltk import CFG

grammar = CFG.fromstring("""
S -> NP VP
NP -> 'John' | 'Mary'
VP -> 'runs' | 'sleeps'
""")

parser = nltk.EarleyChartParser(grammar)
sentence = "John runs".split()

for tree in parser.parse(sentence):
    print(tree)
```
