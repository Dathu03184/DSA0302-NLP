```python
words = ["the", "boy", "plays", "cricket"]

# Initial tags
tags = ["NN", "NN", "NN", "NN"]

# Transformation rule:
# Change a word ending in 's' to VBZ
for i, word in enumerate(words):
    if word.endswith("s"):
        tags[i] = "VBZ"

for word, tag in zip(words, tags):
    print(word, "->", tag)
```
