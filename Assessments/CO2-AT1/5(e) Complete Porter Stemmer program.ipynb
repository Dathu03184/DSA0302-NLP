```python

# Q5: Porter Stemmer-Based Document Retrieval

from nltk.stem import PorterStemmer

words = ["relational", "relation", "relate"]
porter = PorterStemmer()

print(f"{'Original':<15}{'Porter Stem':<15}")
print("-" * 30)

for word in words:
    stem = porter.stem(word)
    print(f"{word:<15}{stem:<15}")
```
