```python
words = ["writes", "writing", "written"]

rules = {
    "writes": ("write + -s", "Regular", "write"),
    "writing": ("write + -ing (drop final e)", "Regular", "write"),
    "written": ("Irregular past participle of write",
                "Irregular", "write")
}

for word in words:
    structure, pattern, root = rules[word]
    print("Word:", word)
    print("Morphology:", structure)
    print("Pattern:", pattern)
    print("Root:", root)
    print("Normalized:", root)
    print("-" * 35)
```
