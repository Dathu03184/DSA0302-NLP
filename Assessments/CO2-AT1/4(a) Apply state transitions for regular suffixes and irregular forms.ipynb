```python
words = ["writes", "writing", "written"]

def transition(word):
    if word == "writes":
        return "q0 -> q1 -> q2 -> q3", "Regular -s"
    elif word == "writing":
        return "q0 -> q1 -> q2 -> q3", "Regular -ing"
    elif word == "written":
        return "q0 -> q1 -> q2 -> q3", "Irregular form"

for word in words:
    path, pattern = transition(word)
    print(word, "->", path, "|", pattern)
```
