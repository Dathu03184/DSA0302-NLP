```python
words = ["writes", "writing", "written"]

transitions = {
    ("q0", "input"): "q1",
    ("q1", "recognized"): "q2",
    ("q2", "valid"): "q3"
}

rules = {
    "writes": ("write + -s", "Regular"),
    "writing": ("write + -ing (drop final e)", "Regular"),
    "written": ("irregular past participle", "Irregular")
}

for word in words:
    state = "q0"
    path = [state]

    state = transitions[(state, "input")]
    path.append(state)

    state = transitions[(state, "recognized")]
    path.append(state)

    state = transitions[(state, "valid")]
    path.append(state)

    structure, pattern = rules[word]

    print("Word:", word)
    print("State path:", " -> ".join(path))
    print("Morphology:", structure)
    print("Pattern:", pattern)
    print("Root: write")
    print("Normalized: write")
    print("-" * 40)
```
