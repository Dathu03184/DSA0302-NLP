```python
words = ["played", "player", "playing"]

def classify(word):
    if word == "played":
        return "Past tense", "Inflectional"
    elif word == "player":
        return "Agent noun", "Derivational"
    elif word == "playing":
        return "Present participle/gerund", "Inflectional"

for word in words:
    form, kind = classify(word)
    print(word, "->", form, "|", kind)
```
