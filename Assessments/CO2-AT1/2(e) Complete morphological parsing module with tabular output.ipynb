```python
words = ["unhappy", "happiness", "happily"]

rules = {
    "unhappy": ("un-", "happy", "None", "Derivational"),
    "happiness": ("None", "happy", "-ness", "Derivational"),
    "happily": ("None", "happy", "-ly", "Derivational")
}

print(f"{'Word':<13}{'Prefix':<10}{'Base':<10}"
      f"{'Suffix':<10}{'Type':<15}{'Root':<10}")
print("-" * 68)

for word in words:
    prefix, base, suffix, kind = rules[word]
    print(f"{word:<13}{prefix:<10}{base:<10}"
          f"{suffix:<10}{kind:<15}{base:<10}")
```
