```python
words = ["played", "player", "playing"]

def remove_affix(word):
    if word == "played":
        return "play", "-ed"
    elif word == "player":
        return "play", "-er"
    elif word == "playing":
        return "play", "-ing"

for word in words:
    stem, affix = remove_affix(word)
    print(word, "-> Stem:", stem, "| Removed:", affix)
```
