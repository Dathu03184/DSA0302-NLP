```python
words = ["played", "player", "playing"]

results = {
    "played": ("play", "-ed", "Inflectional"),
    "player": ("play", "-er", "Derivational"),
    "playing": ("play", "-ing", "Inflectional")
}

for word in words:
    stem, affix, kind = results[word]
    print("Word:", word)
    print("Stem:", stem)
    print("Affix:", affix)
    print("Classification:", kind)
    print("-" * 30)
```
