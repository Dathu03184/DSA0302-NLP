```python
words = ["relational", "relation", "relate"]

for word in words:
    if word == "relational":
        intermediate = "relate"
        stem = "relat"
    elif word == "relation":
        intermediate = "relat"
        stem = "relat"
    else:
        intermediate = "relate"
        stem = "relat"

    print(word, "-> Intermediate:", intermediate,
          "-> Final stem:", stem)
```
