```python
words = ["played", "player", "playing"]

for word in words:
    if word.endswith("ed"):
        root = word[:-2]
    elif word.endswith("er"):
        root = word[:-2]
    elif word.endswith("ing"):
        root = word[:-3]
    else:
        root = word

    print(word, "->", root)
```
