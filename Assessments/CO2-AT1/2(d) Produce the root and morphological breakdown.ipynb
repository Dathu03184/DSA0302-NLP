```python
words = ["unhappy", "happiness", "happily"]

parsed = {
    "unhappy": ("un- + happy", "happy"),
    "happiness": ("happi + ness (y changes to i)", "happy"),
    "happily": ("happi + ly (y changes to i)", "happy")
}

for word in words:
    structure, root = parsed[word]
    print("Word:", word)
    print("Breakdown:", structure)
    print("Root:", root)
    print("-" * 30)
```
