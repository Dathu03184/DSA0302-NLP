```python
words = ["relational", "relation", "relate"]

results = {
    "relational": ("relate", "relat"),
    "relation": ("relat", "relat"),
    "relate": ("relate", "relat")
}

for word in words:
    intermediate, stem = results[word]
    print("Original:", word)
    print("Intermediate:", intermediate)
    print("Final stem:", stem)
    print("-" * 30)
```
