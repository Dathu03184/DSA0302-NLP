```python
words = ["relational", "relation", "relate"]

def explain_rule(word):
    if word == "relational":
        return "-ational -> -ate, then remove -e"
    elif word == "relation":
        return "Remove -ion under the applicable Porter condition"
    else:
        return "Remove final -e under the applicable Porter condition"

for word in words:
    print(word, "->", explain_rule(word))
```
