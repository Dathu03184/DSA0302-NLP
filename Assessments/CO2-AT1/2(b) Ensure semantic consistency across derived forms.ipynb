```python
words = ["unhappy", "happiness", "happily"]

meanings = {
    "unhappy": "Negative feeling",
    "happiness": "State of being happy",
    "happily": "In a happy manner"
}

for word in words:
    print(word, "-> Root: happy | Meaning:", meanings[word])
```
