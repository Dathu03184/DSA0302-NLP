```python
words = ["unhappy", "happiness", "happily"]

rules = {
    "unhappy": ("un-", "happy", "None"),
    "happiness": ("None", "happy", "-ness"),
    "happily": ("None", "happy", "-ly")
}

for word in words:
    prefix, base, suffix = rules[word]
    print(word, "-> Prefix:", prefix,
          "| Base:", base, "| Suffix:", suffix)
```
