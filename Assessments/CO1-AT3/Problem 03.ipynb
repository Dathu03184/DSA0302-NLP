```python
def form(s, f):
    v = "aeiou"
    if f in ("N+PL", "V+3SG"):
        if s.endswith(("s", "x", "z", "ch", "sh")):
            return s + "es"
        if s.endswith("y") and len(s) > 1 and s[-2] not in v:
            return s[:-1] + "ies"
        if s.endswith("o") and len(s) > 1 and s[-2] not in v:
            return s + "es"
        return s + "s"
    if f == "V+PROG":
        if s.endswith("e") and len(s) >= 3 and s[-2] not in v:
            return s[:-1] + "ing"
        if len(s) >= 3 and s[-3] not in v and s[-2] in v and s[-1] not in "wxy" and sum(1 for i in range(len(s)) if s[i] in v and (i == 0 or s[i-1] not in v)) == 1:
            return s + s[-1] + "ing"
        return s + "ing"
    suffix = {"V+PAST": "ed", "ADJ+COMP": "er", "ADJ+SUP": "est"}[f]

    if s.endswith("e"):
        return s + suffix[1:]
    if s.endswith("y") and len(s) > 1 and s[-2] not in v:
        return s[:-1] + "i" + suffix
    if len(s) >= 3 and s[-3] not in v and s[-2] in v and s[-1] not in "wxy" and sum(1 for i in range(len(s)) if s[i] in v and (i == 0 or s[i-1] not in v)) == 1:
        return s + s[-1] + suffix
    return s + suffix
L = int(input())
d = {}
for _ in range(L):
    s, f, w = input().split()
    d[(s, f)] = w
Q = int(input())
for _ in range(Q):
    s, f = input().split()
    print(d[(s, f)] if (s, f) in d else form(s, f))
```
