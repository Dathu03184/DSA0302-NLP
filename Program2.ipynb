```python
from difflib import get_close_matches

dictionary = [
    "apple", "banana", "computer", "python", "programming",
    "language", "science", "student", "college", "university",
    "machine", "learning", "artificial", "intelligence",
    "data", "analysis", "software", "network", "database"
]

word = input("Enter a word: ").lower()

if word in dictionary:
    print(word, "is spelled correctly.")
else:
    suggestions = get_close_matches(word, dictionary, n=3, cutoff=0.5)

    print(word, "is misspelled.")

    if suggestions:
        print("Suggestions:", suggestions)
    else:
        print("No suggestions found.")
```
