```python
import nltk
from nltk.tokenize import sent_tokenize, word_tokenize
from collections import Counter

nltk.download('punkt')
nltk.download('punkt_tab')

paragraph = """Python is a popular programming language.
Python is easy to learn and widely used in data science.
Data science helps us analyze data and find useful information."""

sentences = sent_tokenize(paragraph)

print("A) Sentences:")
for sentence in sentences:
    print(sentence)

words = word_tokenize(paragraph)

print("\nB) Words:")
print(words)

clean_words = [word.lower() for word in words if word.isalpha()]

print("\nC) Total number of words:")
print(len(clean_words))

word_frequency = Counter(clean_words)

print("\nD) Most frequent words:")
for word, count in word_frequency.most_common(5):
    print(word, ":", count)
```
